```python
import numpy as np
import matplotlib.pyplot as plt

def plot(fx, x):
    plt.plot(x,fx,'-')
    plt.axhline(y=0, linewidth=2, color='r')
    plt.axvline(x=0, linewidth=2, color='r')
    plt.grid(True)
    plt.xlabel('X')
    plt.ylabel('Fungsi')
    plt.show()
    
    
```

```python
# Nur Hazbiy Shaffan

x = np.arange(-3,3,1)
fx = x-1
plot(fx,x)

x = np.arange(-1,1,0.1)
fx = x**2-1
plot(fx,x)
```

## Open Method

```python
def fxCalc(x):
    fx = x**7-1000
    return fx
def newton(x):
    xn1 = x-(x**7-1000)/(7*x**6)
    return xn1
def secan(x,xp):
    xn1 = x-(fxCalc(x)*(x-xp))/(fxCalc(x)-fxCalc(xp))
    return xn1
```

```python
import pandas as pd
data = []

x = 3
xn=0
while (fxCalc(x) > 0.000000001):
    xn = newton(x)
    data.append([x, fxCalc(x), newton(x)])
    x=xn
    
data.append([x, fxCalc(x), newton(x)])
pd.DataFrame(data, columns=["Xn", "F(x)","Xn+1"])
```
