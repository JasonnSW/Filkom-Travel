```python

import numpy as np

# Define the function
def f(x):
    return x**2 * np.exp(-x/2) - 1

# Bisection method
def bisection(f, a, b, tol):
    if f(a) * f(b) >= 0:
        print("Bisection method fails.")
        return None
    a_n = a
    b_n = b
    while (b_n - a_n)/2 > tol:
        m_n = (a_n + b_n)/2
        f_m_n = f(m_n)
        if f(a_n) * f_m_n < 0:
            b_n = m_n
        elif f(b_n) * f_m_n < 0:
            a_n = m_n
        elif f_m_n == 0:
            return m_n
        else:
            print("Bisection method fails.")
            return None
    return (a_n + b_n)/2

root_bisection = bisection(f, 0, 2, 1e-3)
root_bisection
```
